```python
## As always, import pandas as ...

import pandas as pd
```

```python
## Read CSV from GitHub url: https://github.com/AlexTheAnalyst/PandasYouTubeSeries/blob/main/Flavors.csv?raw=true

flavors = pd.read_csv("https://github.com/AlexTheAnalyst/PandasYouTubeSeries/blob/main/Flavors.csv?raw=true")
flavors
```

```python
## Explore the data

flavors.head(5)
flavors.shape
flavors.info()
flavors.describe()
## Success? Commit changes to the repository!
```

```python
## Let's group by Base Flavor

flavors['Base Flavor']
flavors['Base Flavor'].count()
flavors['Base Flavor'].value_counts()
flavors.groupby('Base Flavor')['Total Rating'].mean()
flavors.groupby('Base Flavor').mean(numeric_only=True)
```

```python
## Make this a new datafram

group_flavors = flavors.groupby('Base Flavor')

# Calculated the means of the new (Grouped) dataframe. Used the numveric_only argument to make sure that non-numerical values are excluded from calculation.

## Success? Commit changes to the repository!
```

```python
## Let's count the number of flavors for each Base Flavor
group_flavors.size()
```

```python
## Let's get the means for each Base Flavor

group_flavors.mean(numeric_only=True)

## Success? Commit changes to the repository!
```

```python
## Let's try that in one line

```

```python
## Let's discover some popular aggregate statistics like count, sum, min, max. Notice that text columns will appear with min and max as well

group_flavors.min(numeric_only=True)
group_flavors.max(numeric_only=True)
group_flavors.sum(numeric_only=True)
group_flavors.count()

## Success? Commit changes to the repository!
```

```python
flavors.columns
```

```python
## Let's try more advanced aggregation with the `agg` function
## We have to specify which columns we want to aggregate and which functions to apply to them via a dictionary
# group_flavors['Total Rating'].agg({'mean', 'sum','max','min'})

agg_dictionary = {
    "Flavor Rating" : ['mean', 'sum', 'max', 'min'],
    "Texture Rating" : ['mean', 'sum', 'max', 'min']
}

group_flavors.agg(agg_dictionary)
## Success? Commit changes to the repository!
```

```python
## Let's group on multiple columns
## Check the dataframe again and group by Base Flavor and Liked (and show mean as value)
# Per Base Flavor, what is the average Flavor Rating among Liked 'yes' 'no'

flavors.groupby(['Base Flavor','Liked']).mean(numeric_only=True)

# Grouped the original dataframe(flavors) by Base Flavor and Liked. Then, calculated the mean. Got an error,
# Because of strings. Then, added the numeric_only = True argument
```

```python
# Do the multiple aggregation for Base Flavor and Liked

group_by_FR_Liked = flavors.groupby(['Base Flavor','Liked'])
group_by_FR_Liked.agg(agg_dictionary)

## Question from student : Does it matter (for this example) if you change the order of Goupby Columns 
## Answer : compare the table above and below
```
